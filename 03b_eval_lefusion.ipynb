# 03b – Background / mask tests

Same trained model, same patches, same histogram and same seed in every test. Only the background and the generation mask change:

1. noisy real + actual mask
2. noisy real + dilated mask
3. noisy pseudo-healthy + actual mask
4. noisy pseudo-healthy + dilated mask

*Pseudo-healthy* = the aneurysm and its blurred rim (`dilate_voxels`) removed, parent vessel kept, filled with tissue.
*Dilated mask* = actual mask + `dilate_voxels`, without the parent vessel.

Settings: `configs/lefusion_eval_b.yaml`. Results: `outputs/<run_name>/eval_b/<checkpoint>_<split>/`.
Look at `0_pseudo_healthy_check.png` first: if the aneurysm is not cleanly removed, tests 3 and 4 mean nothing.
